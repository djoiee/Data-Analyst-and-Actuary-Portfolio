```python
# HR EMPLOYEE TURNOVER ANALYSIS

"""
PROJECT:
HR Employee Turnover Analysis

OBJECTIVE:
Analyze employee turnover patterns and identify factors
that may affect employee retention, workload, satisfaction,
and overall employee performance.

TOOLS USED:
- Python
- pandas
- matplotlib
- seaborn

AREAS ANALYZED:
- Employee satisfaction
- Workload and project distribution
- Salary levels
- Promotions
- Employee retention
"""

# 1. IMPORT LIBRARIES

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. APPLY VISUAL STYLE

# Applying a dark-themed style for cleaner and modern-looking visualizations
plt.style.use('dark_background')

# Custom color palette
sns.set_palette(['#3B82F6', '#FF007F'])

# 3. LOAD DATASET

# Loading the HR dataset for employee turnover analysis
df = pd.read_csv("HR_comma_sep.csv")

# 4. INITIAL DATA EXPLORATION

print("====================================================")
print("FIRST 5 ROWS OF THE DATASET")
print("====================================================")
print(df.head())

print("\n====================================================")
print("DATASET INFORMATION")
print("====================================================")
print(df.info())

print("\n====================================================")
print("STATISTICAL SUMMARY")
print("====================================================")
print(df.describe())

# 5. DATA QUALITY CHECK

# Checking for missing values to ensure complete analysis
print("\n====================================================")
print("MISSING VALUES")
print("====================================================")
print(df.isnull().sum())

# Identifying duplicate records
print("\n====================================================")
print("DUPLICATE ROWS")
print("====================================================")
print(df.duplicated().sum())

# Removing duplicate rows for cleaner analysis
df.drop_duplicates(inplace=True)

print("\nDuplicate rows removed successfully.")

# 6. EMPLOYEE TURNOVER ANALYSIS

# Analyzing how many employees stayed versus left the company
print("\n====================================================")
print("EMPLOYEE TURNOVER COUNT")
print("====================================================")
print(df['left'].value_counts())

plt.figure(figsize=(7,5))

sns.countplot(data=df, x='left')

plt.title("Employee Turnover")
plt.xlabel("Stayed (0) vs Left (1)")
plt.ylabel("Number of Employees")

plt.show()

print("""
BUSINESS INSIGHT:
A noticeable number of employees left the company,
which may indicate concerns related to workload,
employee satisfaction, or career growth opportunities.
""")

# 7. EMPLOYEE SATISFACTION ANALYSIS

# Analyzing the distribution of employee satisfaction levels
plt.figure(figsize=(8,5))

sns.histplot(df['satisfaction_level'], bins=20)

plt.title("Employee Satisfaction Distribution")
plt.xlabel("Satisfaction Level")
plt.ylabel("Number of Employees")

plt.show()

print("""
BUSINESS INSIGHT:
Employees with lower satisfaction levels appear
more likely to leave the company.
""")

# 8. WORK HOURS ANALYSIS

# Analyzing whether longer work hours are related to employee turnover
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='left', y='average_montly_hours')

plt.title("Work Hours vs Employee Turnover")
plt.xlabel("Stayed (0) vs Left (1)")
plt.ylabel("Average Monthly Hours")

plt.show()

print("""
BUSINESS INSIGHT:
Employees who left the company generally worked
longer hours, suggesting workload may contribute
to employee turnover.
""")

# 9. PROJECT LOAD ANALYSIS

# Evaluating how project workload affects employee retention
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='left', y='number_project')

plt.title("Project Load vs Employee Turnover")
plt.xlabel("Stayed (0) vs Left (1)")
plt.ylabel("Number of Projects")

plt.show()

print("""
BUSINESS INSIGHT:
Employees handling a higher number of projects
were more likely to leave the company.
""")

# 10. SALARY ANALYSIS

# Organizing salary categories from low to high for better analysis flow
df['salary'] = pd.Categorical(
    df['salary'],
    categories=['low', 'medium', 'high'],
    ordered=True
)

# Analyzing the relationship between salary levels and employee turnover
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='salary', hue='left')

plt.title("Salary Level and Employee Turnover")
plt.xlabel("Salary Level")
plt.ylabel("Number of Employees")

plt.show()

print("""
BUSINESS INSIGHT:
Employees with lower salary levels showed
higher turnover rates compared to higher-paid employees.
""")

# 11. YEARS IN COMPANY ANALYSIS

# Analyzing whether years spent in the company affect turnover
plt.figure(figsize=(8,5))

sns.boxplot(data=df, x='left', y='time_spend_company')

plt.title("Years in Company vs Employee Turnover")
plt.xlabel("Stayed (0) vs Left (1)")
plt.ylabel("Years in Company")

plt.show()

print("""
BUSINESS INSIGHT:
Some employees who stayed longer in the company
still decided to leave, which may suggest concerns
related to burnout or career progression.
""")

# 12. PROMOTION ANALYSIS

# Evaluating the relationship between promotions and employee retention
plt.figure(figsize=(8,5))

sns.countplot(data=df, x='promotion_last_5years', hue='left')

plt.title("Promotions and Employee Turnover")
plt.xlabel("Promotion in Last 5 Years")
plt.ylabel("Number of Employees")

plt.show()

print("""
BUSINESS INSIGHT:
Employees without recent promotions appeared
more likely to leave the company.
""")

# 13. CORRELATION ANALYSIS

# Selecting only numerical columns for correlation analysis
numeric_df = df.select_dtypes(include=['int64', 'float64'])

# Visualizing relationships between numerical variables
plt.figure(figsize=(10,6))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Heatmap")

plt.show()

print("""
BUSINESS INSIGHT:
The heatmap highlights relationships between employee
satisfaction, workload, evaluations, and turnover trends.
""")

# 14. FINAL BUSINESS INSIGHTS

print("""
FINAL BUSINESS INSIGHTS

1. Lower employee satisfaction is linked to higher turnover.

2. Employees working longer hours tend to leave more often.

3. Heavy project workloads may contribute to burnout.

4. Lower salary levels show higher resignation rates.

5. Lack of promotion opportunities may affect retention.


BUSINESS RECOMMENDATIONS


1. Improve employee engagement and satisfaction initiatives.

2. Balance project distribution to reduce employee burnout.

3. Monitor employee workloads more effectively.

4. Review compensation and retention strategies.

5. Create clearer career growth and promotion opportunities.
""")

# END OF PROJECT
```
